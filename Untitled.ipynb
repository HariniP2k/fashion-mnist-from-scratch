```python
from tensorflow.keras.datasets import fashion_mnist

(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

print(X_train.shape)
```

```python
X_train = X_train.reshape(60000, 784) / 255.0
X_test = X_test.reshape(10000, 784) / 255.0
```

```python
import numpy as np

def one_hot(y, num_classes=10):
    one_hot_y = np.zeros((y.size, num_classes))
    one_hot_y[np.arange(y.size), y] = 1
    return one_hot_y

y_train_encoded = one_hot(y_train)
y_test_encoded = one_hot(y_test)

print(y_train_encoded.shape)
```

```python
def forward(X):
    global Z1, A1, Z2, A2

    Z1 = np.dot(X, W1) + b1
    A1 = relu(Z1)

    Z2 = np.dot(A1, W2) + b2
    A2 = softmax(Z2)

    return A2
```

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)
```

```python
def cross_entropy(y_true, y_pred):
    epsilon = 1e-10
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    return -np.mean(np.sum(y_true * np.log(y_pred), axis=1))
def mse_loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

```python
input_size = 784
hidden_size = 128
output_size = 10

W1 = np.random.randn(input_size, hidden_size) * np.sqrt(2. / input_size)
b1 = np.zeros((1, hidden_size))

W2 = np.random.randn(hidden_size, output_size) * np.sqrt(2. / hidden_size)
b2 = np.zeros((1, output_size))
```

```python
sW1 = np.zeros_like(W1)
sb1 = np.zeros_like(b1)

sW2 = np.zeros_like(W2)
sb2 = np.zeros_like(b2)

def backward(X, y_true, learning_rate):

    global W1, b1, W2, b2
    global sW1, sb1, sW2, sb2

    m = X.shape[0]

    # Output layer gradient
    dZ2 = A2 - y_true

    dW2 = np.dot(A1.T, dZ2) / m
    db2 = np.sum(dZ2, axis=0, keepdims=True) / m

    # Hidden layer gradient
    dA1 = np.dot(dZ2, W2.T)

    dZ1 = dA1 * (Z1 > 0)

    dW1 = np.dot(X.T, dZ1) / m
    db1 = np.sum(dZ1, axis=0, keepdims=True) / m

    # RMSProp Update
    beta = 0.9
    epsilon = 1e-8

    sW1 = beta * sW1 + (1 - beta) * (dW1 ** 2)
    sb1 = beta * sb1 + (1 - beta) * (db1 ** 2)

    sW2 = beta * sW2 + (1 - beta) * (dW2 ** 2)
    sb2 = beta * sb2 + (1 - beta) * (db2 ** 2)

    W1 -= learning_rate * dW1 / (np.sqrt(sW1) + epsilon)
    b1 -= learning_rate * db1 / (np.sqrt(sb1) + epsilon)

    W2 -= learning_rate * dW2 / (np.sqrt(sW2) + epsilon)
    b2 -= learning_rate * db2 / (np.sqrt(sb2) + epsilon)
```

```python
def accuracy(y_true, y_pred):
    y_true_labels = np.argmax(y_true, axis=1)
    y_pred_labels = np.argmax(y_pred, axis=1)

    return np.mean(y_true_labels == y_pred_labels)
```

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)
```

```python
from sklearn.model_selection import train_test_split

X_train_split, X_val, y_train_split, y_val = train_test_split(
    X_train,
    y_train_encoded,
    test_size=0.1,
    random_state=42
)

print(X_train_split.shape)
print(X_val.shape)
```

```python
epochs = 5
learning_rate = 0.001
batch_size = 64

for epoch in range(epochs):

    indices = np.random.permutation(X_train_split.shape[0])

    X_shuffled = X_train_split[indices]
    y_shuffled = y_train_split[indices]

    for i in range(0, X_train_split.shape[0], batch_size):

        X_batch = X_shuffled[i:i+batch_size]
        y_batch = y_shuffled[i:i+batch_size]

        predictions = forward(X_batch)

        loss = mse_loss(y_batch, predictions)

    

        backward(X_batch, y_batch, learning_rate)

    train_predictions = forward(X_train_split)

train_acc = accuracy(y_train_split, train_predictions)

val_predictions = forward(X_val)

val_acc = accuracy(y_val, val_predictions)

wandb.log({
    "loss": loss,
    "train_accuracy": train_acc,
    "validation_accuracy": val_acc
})

print(f"Epoch {epoch+1}")
print(f"Loss: {loss:.4f}")
print(f"Accuracy: {train_acc:.4f}")
print(f"Validation Accuracy: {val_acc:.4f}")
```

```python
sweep_config = {
    'method': 'random',
    'metric': {
        'name': 'validation_accuracy',
        'goal': 'maximize'
    },
    'parameters': {
        'learning_rate': {
            'values': [0.001, 0.0001]
        },
        'batch_size': {
            'values': [32, 64]
        },
        'epochs': {
            'values': [5]
        }
    }
}
```

```python
X_val = X_train[:6000]
y_val = y_train[:6000]

X_train = X_train[6000:]
y_train = y_train[6000:]

y_train_encoded = one_hot(y_train)
y_val_encoded = one_hot(y_val)
```

```python
def train():

    wandb.init()

    config = wandb.config

    learning_rate = config.learning_rate
    batch_size = config.batch_size
    epochs = config.epochs

    for epoch in range(epochs):

        indices = np.random.permutation(X_train.shape[0])

        X_shuffled = X_train[indices]
        y_shuffled = y_train_encoded[indices]

        for i in range(0, X_train.shape[0], batch_size):

            X_batch = X_shuffled[i:i+batch_size]
            y_batch = y_shuffled[i:i+batch_size]

            predictions = forward(X_batch)

            loss = cross_entropy(y_batch, predictions)

            backward(X_batch, y_batch, learning_rate)

        train_predictions = forward(X_train)

        train_acc = accuracy(y_train_encoded, train_predictions)

        val_predictions = forward(X_val)

        val_acc = accuracy(y_val_encoded, val_predictions)

        wandb.log({
            "loss": loss,
            "train_accuracy": train_acc,
            "validation_accuracy": val_acc
        })
```

```python
sweep_id = wandb.sweep(sweep_config, project="fashion-mnist")
```

```python
wandb.agent(sweep_id, function=train, count=3)
```
